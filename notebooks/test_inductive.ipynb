```python
import dgl
import torch

g_path = '/root/autodl-tmp/source/pprogo-flg/data/bp/graph.dgl'
g, _ = dgl.load_graphs(str(g_path))
g = g[0]
```

```python
nids = [1,2,3]
# 'go_annotation':range(g.number_of_nodes('go_annotation'))
subgraph = dgl.in_subgraph(g, {'protein':nids}, relabel_nodes=True)
print(subgraph)
```

```python
go_graph = dgl.node_subgraph(g, {'go_annotation':range(g.number_of_nodes('go_annotation'))})
print(go_graph)
```

```python
test_graph = dgl.merge([subgraph, go_graph])
print(test_graph)
```
